# 梯度下降算法

```python
import matplotlib.pyplot as plt
```

```python
x_data = [1.0,2.0,3.0]
y_data = [2.0,4.0,6.0]

l = min(len(x_data),len(y_data))

w=1.0 # 先猜测一个模型参数w

def forward(x):
    # 选择不带参数b的模型简化
    return x*w 

# 计算损失函数
def cost(xs,ys):
    cost = 0
    for x,y in zip(xs,ys):
        y_pred = forward(x)
        cost += (y_pred-y)**2
    return cost/len(xs)

# 计算梯度
def gradient(xs,ys):
    grad = 0
    for x,y in zip(xs,ys):
        grad += 2*x*(x*w-y) # 梯度计算公式由cost对w求偏导得到
    return grad/len(xs)
```

```python
print("训练前测试x及其预测值y",4,forward(4))
```

```python
w_list = []
mse_list = []
for epoch in range(100): # 训练99轮,epoch表示轮次,epoch=0相当于还没训练过
    cost_val = cost(x_data,y_data)
    gral_val = gradient(x_data,y_data)
    w = w - 0.01 * gral_val # 学习率设为0.01,根据梯度值更新参数w
    print("epoch=",epoch,"w=",w,"cost_val=",cost_val)
    w_list.append(w)
    mse_list.append(cost_val)
print("训练后测试x及其预测值y",4,forward(4))
```

```python
plt.xlabel("epoch")
plt.ylabel("MSE")
plt.plot([epoch for epoch in range(0,100)],mse_list)
plt.show()
```

## 使用随机梯度下降

```python
x_data = [1.0,2.0,3.0]
y_data = [2.0,4.0,6.0]

l = min(len(x_data),len(y_data))

w=1.0 # 先猜测一个模型参数w

def forward(x):
    # 选择不带参数b的模型简化
    return x*w 

# 计算损失函数
def loss(x,y):
    pred_y = forward(x)
    return (pred_y-y)**2

# 计算新梯度公式
def gradient(x,y):
    return 2*x*(x*w-y)

for epoch in range(100):
    for x,y in zip(x_data,y_data):
        grad = gradient(x,y) # 理论上这里应该把元组重排实现随机
        w = w - 0.01*grad
        print("\tgrad:",x,y,grad)
        l = loss(x,y)
    print("progress:",epoch,"w=",w,"loss=",l)
```
