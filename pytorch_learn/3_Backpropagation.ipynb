# 反向传播 

```python
import torch
x_data = [1.0,2.0,3.0]
y_data = [2.0,4.0,6.0]
w = torch.Tensor([1.0]) # 创建一个1.0的一维张量
w.requires_grad = True # 设置需要计算w的梯度
def forward(x):
    return x*w # w是Tensor对象，此时*是矩阵数乘，x自动转化为Tensor
def loss(x,y):
    pred_y = forward(x)
    return (pred_y-y)**2

print("predict (before training)", 4, forward(4).item())
# item()将单元素张量转为Python标量
```

```python
for epoch in range(100):
    for x, y in zip(x_data, y_data):
        l = loss(x, y)
        l.backward() # backward()方法计算梯度，存入w.grad并释放计算图
        print("\tgrad:", x, y, w.grad.item())

        with torch.no_grad():
            w -= 0.01*w.grad # 更新参数，但这一步不要计算梯度

        w.grad.zero_() # backward()计算得到的梯度会积累，用zero_()清零

    print("progress:", epoch, l.item())

print("predict (after training)", 4, forward(4).item())
```
