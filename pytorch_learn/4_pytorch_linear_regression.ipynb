# pytorch中构建线性回归模型y_hat=wx+b

```python
import torch
x_data = torch.tensor([[1.],[2.],[3.]])
y_data = torch.tensor([[2.],[4.],[6.]])

# 定义线性模型的类，继承自torch.nn.Module（所有神经网络的父类）
class LinearModel(torch.nn.Module): 
    def __init__(self): # self可理解为模型本身
        super(LinearModel, self).__init__() # 先执行父类 torch.nn.Module 的初始化
        self.linear = torch.nn.Linear(1, 1) # nn.Linear表示对于y = Wx + b的线性模型
        # (1,1)表示输入x是1维度，输出y是1维度，默认bias = True
        # w和b已经自动随机生成
        # 将linear这个线性模型加入到模型中

    def forward(self, x): # 定义前向传播
        y_pred = self.linear(x)
        return y_pred

model = LinearModel() # 实例化
criterion = torch.nn.MSELoss(reduction="sum") # nn.MSELoss也是继承自nn.Module
# 定义损失函数，平方误差求和（sum）（不平均）
optimizer = torch.optim.SGD(model.parameters(), lr=0.01) # 优化器，lr为学习率
# 告诉优化器要优化的参数，以及学习率，model.parameters()返回所有参数

for epoch in range(100):
    y_pred = model(x_data)
    loss = criterion(y_pred, y_data)
    print("epoch:", epoch, "loss:", loss)

    optimizer.zero_grad() # 梯度清零
    loss.backward()
    optimizer.step() # 更新参数

print("w=", model.linear.weight.item())
print("b=", model.linear.bias.item())

x_test = torch.tensor([[4.0]])
y_test = model(x_test)
print("predict (after training)", 4, y_test)
```
