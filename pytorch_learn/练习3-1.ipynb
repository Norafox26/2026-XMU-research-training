# 用二次多项式作为模型训练

```python
import torch

x_data = [1.0, 2.0, 3.0]
y_data = [2.0, 4.0, 6.0]

w1 = torch.tensor([1.0], requires_grad=True)
w2 = torch.tensor([1.0], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

# 前向传播
def forward(x):
    return w1 * x**2 + w2 * x + b

# 损失函数
def loss(x, y):
    y_pred = forward(x)
    return (y_pred - y)**2

# 训练
for epoch in range(1000):
    for x, y in zip(x_data, y_data):
        l = loss(x, y)
        l.backward() # 反向传播，计算三个参数的梯度
        # 更新参数
        with torch.no_grad():
            w1 -= 0.01 * w1.grad
            w2 -= 0.01 * w2.grad
            b -= 0.01 * b.grad
        # 梯度清零
        w1.grad.zero_()
        w2.grad.zero_()
        b.grad.zero_()

    if epoch % 100 == 0:
        print(
            "epoch:", epoch,
            "loss:", l.item(),
            "w1:", w1.item(),
            "w2:", w2.item(),
            "b:", b.item()
        )# 每100轮输出一次结果

print("\n训练结果：")
print("w1 =", w1.item())
print("w2 =", w2.item())
print("b  =", b.item())

print("x=4时预测 =", forward(4).item())
```
