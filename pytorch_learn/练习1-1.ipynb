# 以y=wx+b为模型，实现线性回归

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_data = [1.0,2.0,3.0,4.0]
y_data = [3.0,5.0,7.0,9.0]
l = min(len(x_data),len(y_data))
def forward(x):
    return x*w+b
def loss(x,y):
    y_pred = forward(x)
    return (y_pred-y)**2
```

```python
w_list = []
b_list = []
mse_list = []
for w in np.arange(-1,5.1,0.1):
    for b in np.arange(-1,5.1,0.1):
        loss_sum = 0
        for x,y in zip(x_data,y_data):
            y_val_pred = w*x+b
            loss_sum += loss(x,y)
            print("\t",x,y,y_val_pred,loss(x,y))
        print("MSE=",loss_sum/l)
        w_list.append(w)
        b_list.append(b)
        mse_list.append(loss_sum/l)
```

```python
n_w = len(np.unique(w_list)) # np.unique()找出列表中不重复的元素的个数
n_b = len(np.unique(b_list))

W = np.array(w_list).reshape(n_w, n_b) # np.array()将列表转换成numpy一维数组；reshape(n_w, n_b)按照61*61转换成矩阵
B = np.array(b_list).reshape(n_w, n_b)
MSE = np.array(mse_list).reshape(n_w, n_b)

fig = plt.figure() # 创建一个figure对象保存在fig
ax = fig.add_subplot(111, projection='3d') # 在fig创建一个3D的坐标轴对象ax，111表示在fig中第一行第一列画第一个子图

ax.plot_surface(W, B, MSE) # plot_surface()根据三维散点绘制曲面

ax.set_xlabel('w')
ax.set_ylabel('b')
ax.set_zlabel('MSE')

plt.show()
```
